```python
import pickle

with open('/data/ephemeral/home/level2-objectdetection-cv-01/mmdetection/submission.pkl', 'rb') as file:
    data = pickle.load(file)

print(data)
```

```python
import pandas as pd

# 결과를 저장할 리스트 초기화
prediction_strings = []
file_names = []

# Pickle 데이터를 순회하면서 필요한 정보 추출
for item in data:
    prediction_string = ''
    # 각 클래스별 bounding box, score, label 정보 추출 및 포맷팅
    for bbox, score, label in zip(item['pred_instances']['bboxes'], item['pred_instances']['scores'], item['pred_instances']['labels']):
        prediction_string += f"{label.item()} {score.item()} {bbox[0].item()} {bbox[1].item()} {bbox[2].item()} {bbox[3].item()} "

    prediction_strings.append(prediction_string.strip())
    file_names.append('test/' + item['img_path'].split('/')[-1])  # 경로에서 파일 이름만 추출

# DataFrame 생성 및 CSV 파일로 저장
submission = pd.DataFrame({
    'PredictionString': prediction_strings,
    'image_id': file_names
})

# CSV 파일 저장
submission.to_csv('submission.csv', index=False)
```
